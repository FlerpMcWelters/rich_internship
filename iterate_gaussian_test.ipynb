```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
#parameters
n = 1000000
k = 6346
mu1 = 20000 #mean of distro 1
mu2 = 20500 #mean of distro 2

#threshold value
t = (mu2-mu1)/(np.sqrt(n))

#Bob randomly selects a value of S in range
S_start = 10000
S_end = 10000000

num_trials = 5000

print(t)
```

```python
#Establish success rate across 1000 iterations
outcomecount1 = 0
outcomecount2 = 0
outcomecount3 = 0
outcomecount4 = 0

num_successes = 0
for i in range(num_trials):

#Bob establishes S
    
    S = random.randint(S_start, S_end)
    
#Bob randomly selects set
    
    bobChoice = random.choice(["U","V"])
    
#Establish sets
    
    U = np.random.normal(loc=mu1, scale=1, size=k) #Random sampling of distribution1 - this is public
    V = np.random.normal(loc=mu2, scale=1, size=k) #random sampling of distribution2 - this is public

#Generate random selection of half these values

    choiceU = random.sample(tuple(U), int(k/2))
    choiceV = random.sample(tuple(V), int(k/2))
    
#Sum sets with S
    
    sum_sampleU = sum(choiceU)+S
    sum_sampleV = sum(choiceV)+S
    
#Establish what to send to Alice
    
    if bobChoice == "U":
        aliceSample = sum_sampleU
    elif bobChoice == "V":
        aliceSample = sum_sampleV
    else:
        print("error")

#Alice now flips a coin to pick a distribution

    aliceChoice = random.choice(["Distribution 1","Distribution 2"])

#Alice now generates and adds n-(k/2) samples of chosen distribution to Bob's transmission
    
    if aliceChoice == "Distribution 1":
        x = np.random.normal(loc = mu1, scale=1, size=int(n-(k/2)))
        aliceSum = sum(x) + aliceSample
    elif aliceChoice == "Distribution 2":
        y = np.random.normal(loc=mu2, scale=1, size=int(n-(k/2)))
        aliceSum = sum(y) + aliceSample
    else:
        print("problem")
        
#Now alice divides this value by n: call this A

    A = aliceSum/n

#Alice will send E = A - u to Bob where u is either mu1 or mu2 depending on Alice's choice

    if aliceChoice == "Distribution 1":
        E = A - mu1
    elif aliceChoice == "Distribution 2":
        E = A - mu2
    else:
        print("problem")


    if abs(E-S/n) <= t:#then they agree on the distribution and bob knows shared bit is 1 if he chose d1 and 0 otherwise
        if bobChoice == "U":
            sharedBit = 1
            outcomecount1 += 1
        elif bobChoice == "V":
            sharedBit = 0
            outcomecount2 += 1
            #print(bobChoice)
        else:
            print("problem")
    elif abs(E- S/n) > t: #bob selected a diff distro than alice - bob still knows alice's distribution
        if bobChoice == "U":
            sharedBit=0
            outcomecount3 += 1
            #print(bobChoice)
        elif bobChoice == "V":
            sharedBit = 1
            outcomecount4 += 1
        else:
            print("problem")

#Establish if they selected the same distribution and count number of successes.
    
    if (sharedBit == 1) and (aliceChoice == "Distribution 1"):
        num_successes += 1
        #print("Success")
    elif (sharedBit == 0) and (aliceChoice == "Distribution 2"):
        num_successes += 1
        #print("Success")
    #else:
        #print("failure")

print(num_successes)
print(num_successes/num_trials)

print(outcomecount1)
print(outcomecount2)
print(outcomecount3)
print(outcomecount4)
```
