# Double-transmission protocol

#### This is an effective duplicate of double_transmission_protocol with a random selection between U and V for Bob implemented

In this file, Alice will compute $E_u$ and $E_v$, and send both sums back to Bob. Bob will then make a guess based on which of $E - S/n$ is closer to 0 as to which one represents distribution U and inform Alice to select the bit associated with that transmission.

For example, if Bob believes $E_1$ (public name) to be the $E_u$ distribution, he will select 1 as his bit. He informs Alice of his choice, and if he was right then they agree. If not, Alice will receive back the distribution associated with V and therefore select 0.

- Notice that this protocol diverges from the previous one in that Bob no longer experimentally determines a threshold.

- Distribution U is an exponential distribution and so is V.

- There is a function copy with pareto distributions instead of exponential and shape parameter greater than 1. The mean of a pareto distribution is only defined for shape parameter $\alpha>1$ and in this case is commonly referred to as a Pareto Type I. Here is the formula for the mean of the distribution:
$$\frac{\alpha x_m}{\alpha-1}$$

where $x_m$ is the scale parameter (minimum possible value).

```python
import numpy as np
import pandas as pd
import csv, os, datetime as dt
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
def run_protocol(n=100000, k=8, q=200, mu1=8000, mu2=10000,
                 S_start=1_000_000, S_end=10_000_000,
                 num_trials=2000, seed=None):
    
    """Run one full experiment and return a summary dict (one notebook run = one call).

    A fresh np.random.Generator is created from `seed`, so runs are isolated and
    reproducible. Pass seed=None for a different draw each time.
   
    n          --> number of total samples
    k          --> size of set U
    q          --> size of Bob's selection from U
    mu1        --> mean of distribution U, must be greater than shift
    shift      --> amount by which mu1 is shifted to create V's mean i.e. shift=100 alice chooses btwn mu1+100 and mu1-100
    S_start    --> lower bound for Bob's private key
    S_end      --> upper bound for Bob's private key
    num_trials --> number of iterations the protocol is run for
    seed       --> set seed for reproducibility
    
    """
    rng = np.random.default_rng(seed)
    U = rng.exponential(scale=mu1, size=k)                     # public set, fixed for this run - If you want the same U for every
    V = rng.exponential(scale=mu2, size=k)                                                        # trial, move U to empty cell above this function and ensure you
                                                                                                     # edit the k value.

    successes = correct_U = total_U = correct_V = total_V = 0

    for _ in range(num_trials):

        bob_pick = U if rng.random() < 0.5 else V
        bob_sum = rng.choice(bob_pick, size = q).sum() #bob flips a coin to determine his selection and then oversamples from parent set U or V.
        S = rng.integers(S_start, S_end + 1)
        bob_tx = bob_sum + S                                  # Bob -> Alice (S masks bob_sum)

        
        alice_sum_U = rng.exponential(scale=mu1, size=n - q).sum() + bob_tx
        alice_sum_V = rng.exponential(scale=mu2, size=n-q).sum() + bob_tx #note that Alice creates both sets

        
        E_1 = alice_sum_U/n - mu1 #Alice calculates what to send back to Bob
        E_2 = alice_sum_V/n - mu2 #Alice ---> Bob

        bob_bit = 1 if rng.random() < 0.5 else 0 
        
        if abs(E_1 - S/n) <= abs(E_2 - S/n): #Bob assumes E_1 is the scenario in which Alice selected the same distribution as him
            bob_guess_same = E_1
            bob_guess_diff = E_2
        else:
            bob_guess_same = E_2 #Bob assumes E_2 is the scenario in which Alice selected the same distribution.
            bob_guess_diff = E_1
            
        if bob_bit == 1:
            alice_bit = 1 if bob_guess_same == E_1 else 0
            bob_final_bit = 1 if bob_pick is U else 0
        else:
            alice_bit = 1 if bob_guess_diff == E_1 else 0
            bob_final_bit = 0 if bob_pick is U else 1

        successes += (bob_final_bit == alice_bit)
        

    return {
        "success_rate":       successes / num_trials,
        "S_start":            S_start,
        "S_end":              S_end,
        "S_range":            f"{-S_start+S_end}",
        "q":                  q,
        "n":                  n,
        "mu1":                mu1,
        "mu2":                mu2,
        "num_trials":         num_trials,
        "Set U":              U,
        "Set V":              V,
        "Mean of U":          sum(U)/len(U),
        "Mean of V":          sum(V)/len(V),
        "seed":               seed,
    }
```

```python
def run_protocol_pareto(n=100000, k=8, q=200, shape_U = 2, support_U = 4000, shape_V = 2, support_V = 5000,
                 S_start=1_000_000, S_end=10_000_000,
                 num_trials=2000, seed=None):
    
    """
    Since the distributions are now pareto, slightly different arguments will need to be passed to this function.
    support = min value 
    shape > 1 denotes size of tail (higher shape, smaller tail)
    
    
    
    Run one full experiment and return a summary dict (one notebook run = one call).

    A fresh np.random.Generator is created from `seed`, so runs are isolated and
    reproducible. Pass seed=None for a different draw each time.
   
    n          --> number of total samples
    k          --> size of set U
    q          --> size of Bob's selection from U
    mu1        --> mean of distribution U, must be greater than shift
    shift      --> amount by which mu1 is shifted to create V's mean i.e. shift=100 alice chooses btwn mu1+100 and mu1-100
    S_start    --> lower bound for Bob's private key
    S_end      --> upper bound for Bob's private key
    num_trials --> number of iterations the protocol is run for
    seed       --> set seed for reproducibility
    
    """
    rng = np.random.default_rng(seed)
    U = (rng.pareto(a=shape_U, size=k) + 1) * support_U                                                             # edit the k value.                    # public set, fixed for this run - If you want the same U for every
    V = (rng.pareto(a=shape_V, size=k) + 1) * support_V                                                       # trial, move U to empty cell above this function and ensure you

    expectation_U = (support_U*shape_U)/(shape_U-1)
    expectation_V = (support_V*shape_V)/(shape_V-1)

    successes = correct_U = total_U = correct_V = total_V = 0

    for _ in range(num_trials):

        bob_pick = U if rng.random() < 0.5 else V
        bob_sum = rng.choice(bob_pick, size = q).sum() #bob flips a coin to determine his selection and then oversamples from parent set U or V.
        S = rng.integers(S_start, S_end + 1)
        bob_tx = bob_sum + S                                  # Bob -> Alice (S masks bob_sum)

        
        alice_sum_U = ((rng.pareto(a=shape_U, size=n-q) + 1) * support_U).sum() + bob_tx
        alice_sum_V = ((rng.pareto(a=shape_V, size=n-q) + 1) * support_V).sum() + bob_tx #note that Alice creates both sets

        
        E_1 = alice_sum_U/n - expectation_U #Alice calculates what to send back to Bob
        E_2 = alice_sum_V/n - expectation_V #Alice ---> Bob

        bob_bit = 1 if rng.random() < 0.5 else 0 
        
        if abs(E_1 - S/n) <= abs(E_2 - S/n): #Bob assumes E_1 is the scenario in which Alice selected the same distribution as him
            bob_guess_same = E_1
            bob_guess_diff = E_2
        else:
            bob_guess_same = E_2 #Bob assumes E_2 is the scenario in which Alice selected the same distribution.
            bob_guess_diff = E_1
            
        if bob_bit == 1:
            alice_bit = 1 if bob_guess_same == E_1 else 0
            bob_final_bit = 1 if bob_pick is U else 0
        else:
            alice_bit = 1 if bob_guess_diff == E_1 else 0
            bob_final_bit = 0 if bob_pick is U else 1

        successes += (bob_final_bit == alice_bit)
        

    return {
        "success_rate":       successes / num_trials,
        "S_start":            S_start,
        "S_end":              S_end,
        "S_range":            f"{-S_start+S_end}",
        "q":                  q,
        "n":                  n,
        "theoretical_expectation_U":      expectation_U,
        "theoretical_expectation_V":      expectation_V,
        "num_trials":         num_trials,
        "Set U":              U,
        "Set V":              V,
        "Mean of U":          sum(U)/len(U),
        "Mean of V":          sum(V)/len(V),
        "Last Alice mean":           alice_sum_U/n,
        "Last Alice mean":           alice_sum_V/n,
        "seed":               seed,
    }
```

```python
def append_run(summary, log_path="double_transmission_run_log.csv"):
    """Append one summary dict as a row, writing the header only if the file is new."""
    record = {"timestamp": dt.datetime.now().isoformat(timespec="seconds"), **summary}
    write_header = not os.path.isfile(log_path)
    with open(log_path, "a", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(record.keys()))
        if write_header:
            w.writeheader()
        w.writerow(record)
    return record
```

Eve is an eavesdropper who has access only to public keys and information shared publicly by the communicating parties Bob and Alice. I will add in an eavesdropper to the above protocol.
