```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
'''
Alice and Bob are the two communicating parties. Alice has a public and private key. 

Private Key:

Two diff probability distributions with distinct means 

Public Key:

Set U containing k samples of distro 1
Set V contain k samples of distro 2

A large integer n

A range for S

A threshold number t
'''
n = 100000
k=8
mu1 = 20000 #mean of distro 1
mu2 = 20500 #mean of distro 2
t = (mu1+mu2)/np.sqrt(n)
#Bob randomly selects a value of S in range
S = random.randint(10000,1000000)
print(S)
print(t)
```

```python
#Bob flips a coin to determine which set he chooses
bobChoice = random.choice(["U","V"])
print(bobChoice)
```

```python

#Sample both exponential distributions k times

U = np.random.exponential(scale=mu1, size=k) #Random sampling of distribution1 - this is public
print(U)

V = np.random.exponential(scale=mu2, size=k) #random sampling of distribution2 - this is public
print(V)
```

```python
#Generate random selection of half these values

choiceU = random.sample(tuple(U), int(k/2))
choiceV = random.sample(tuple(V), int(k/2))

print(choiceU)
print(choiceV)
```

```python
#Sum these values together with S

sum_sampleU = sum(choiceU)+S
sum_sampleV = sum(choiceV)+S
print(sum_sampleU)
print(sum_sampleV)

# This value is sent to Alice -

if bobChoice == "U":
    aliceSample = sum_sampleU
elif bobChoice == "V":
    aliceSample = sum_sampleV
else:
    print("error")

print(aliceSample)
```

```python
#Alice now flips a coin to determine which distribution

aliceChoice = random.choice(["Distribution 1","Distribution 2"])
print(aliceChoice)
```

```python
#Alice now generates and adds n-(k/2) samples of chosen distribution to Bob's transmission

if aliceChoice == "Distribution 1":
    x = np.random.exponential(scale=mu1, size=int(n-(k/2)))
    aliceSum = sum(x) + aliceSample
elif aliceChoice == "Distribution 2":
    y = np.random.exponential(scale=mu2, size=int(n-(k/2)))
    aliceSum = sum(y) + aliceSample
else:
    print("problem")
print(aliceSum)
    
```

```python
#Now alice divides this value by n: call this A

sum_over_n = aliceSum/n

print(sum_over_n)
```

```python
"""
This value should be *close* (see threshold parameter) to S/n + u if they selected the same distribution, and 
*far* if they did not
"""
#Alice will send E = A/n - u to Bob where u is either mu1 or mu2 depending on Alice's choice

if aliceChoice == "Distribution 1":
    E = sum_over_n - mu1
elif aliceChoice == "Distribution 2":
    E = sum_over_n - mu2
else:
    print("problem")

print(E)
```

```python
#Bob checks if E is close to S/n. recall that whichever distribution alice selects is the shared bit.

if abs(E-S/n) <= t:#then they agree on the distribution and bob knows shared bit is 1 if he chose d1 and 0 otherwise
    if bobChoice == "U":
        sharedBit = 1
    elif bobChoice == "V":
        sharedBit = 0
    else:
        print("problem")
elif abs(E-S/n) > t: #bob selected a diff distro than alice - bob still knows alice's distribution
    if bobChoice == "U":
        sharedBit=0
    elif bobChoice == "V":
        sharedBit = 1
    else:
        print("problem")

if (sharedBit == 1) and (aliceChoice == "Distribution 1"):
    print("Success")
elif (sharedBit == 0) and (aliceChoice == "Distribution 2"):
    print("Success")
else:
    print("Failure")
print(sharedBit)
print(bobChoice)
print(aliceChoice)
print(abs(E-S/n) <= t)
```
